```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
# Path to the csv
DATA_PATH   = Path('/media/haupert/data/mes_projets/05_xprize/performance')

# list of csv
CSV_LIST    = [
            'XP_final_annotations - rec1dmu.csv',
            'XP_final_annotations - rec3dmu.csv',
            'XP_final_annotations - rec3dmu_v2.csv',
            'XP_final_annotations - rec4dmu.csv',
            'XP_final_annotations - rec4dmu_v2.csv',
            'XP_final_annotations - rec6dmu.csv',
            'XP_final_annotations - rec7dmu.csv',
            ]
```

```python
# Load data in the CSV_LIST
df = pd.DataFrame()
for csv in CSV_LIST:
    df_tempo = pd.read_csv(DATA_PATH / csv)
    # Add the name of the file in the dataframe to keep track of the site
    # keep only the name of the file after the last '-' and before the '.csv'
    df_tempo['site'] = csv.split(' - ')[1].split('.csv')[0]
    df = pd.concat([df, df_tempo], axis=0)

# lower case all strings
df= df.map(lambda s:s.lower() if type(s) == str else s)

# display the total number of clusters per site
display(df.groupby('site')['cluster_number'].nunique())

# # display
# display(df)

# # for each site, keep only the cluster_number where all validator_name are different from 0
# df2 = pd.DataFrame()
# for site in df['site'].unique():
#     df2 = pd.concat([df2, df.loc[df['site'] == site].groupby('cluster_number').filter(lambda x: (x['validated_specie'] != '0').all() and (x['validated_specie'] != 0).all() and (x['validated_specie'] != np.nan).all())], axis=0)

# # display
# display(df2)

#  keep the rows where validated_specie is different from 0 and '0'
df3 = pd.DataFrame()
df3 = df.loc[(df['validated_specie'] != '0') & (df['validated_specie'] != 0) & (df['validated_specie'] != np.nan)]

# remove validator_name with '0' and nan
df3 = df3.loc[(df3['validator_name'] != '0') & (df3['validator_name'] is not None)]

# test if the validated_specie is a string if not, remove the row
df3 = df3.loc[df3['validated_specie'].map(lambda x: type(x) == str)]

# in the validated_specie column
# keep the first string before the ',' if any
df3['validated_specie'] = df3['validated_specie'].map(lambda x: x.split(',')[0] if ',' in x else x)
# keep the first string before the '/' if any
df3['validated_specie'] = df3['validated_specie'].map(lambda x: x.split('/')[0] if '/' in x else x)
# keep the first string before the '-' if any   
df3['validated_specie'] = df3['validated_specie'].map(lambda x: x.split('-')[0] if '-' in x else x)
# keep the first string before the '_' if any
df3['validated_specie'] = df3['validated_specie'].map(lambda x: x.split('_')[0] if '_' in x else x)

# display
display(df3)

# display unique validated_specie
display(df3['validated_specie'].unique())

# display the number of unique validated_specie
print(f'the number of validated specie is {len(df3["validated_specie"].unique())}')
```

```python
from sklearn.metrics.cluster import adjusted_mutual_info_score, adjusted_rand_score, normalized_mutual_info_score
from sklearn.metrics import v_measure_score, homogeneity_score, completeness_score

# add a column predicted_label which is the sum of cluster_number 
df3['predicted_label'] = df3['cluster_number'] 

# # convert period to integer when night -> 10000, dawn -> 20000, day -> 30000, duk -> 40000
# df3['period_int'] = df3['period'].map(lambda x: 10000 if x == '1. night' else 20000 if x == '2. dawn' else 30000 if x == '3. day' else 40000 if x == '4. dusk' else x)

# # add a column predicted_label which is the sum of cluster_number and period_int
# df3['predicted_label'] = df3['cluster_number'] + df3['period_int']

# transform the validated_specie into integer via a dictionary
species = df3['validated_specie'].unique()
species = species[species != '0']
species = species[species != 0]
species = species[species != np.nan]
species = species[species != 'nan']

# create a dictionary to transform the validated_specie into integer
species_dict = {species[i]: i for i in range(len(species))}
df3['true_label'] = df3['validated_specie'].map(species_dict)

# display
display(df3)

# compute the adjusted_mutual_info_score and adjusted_rand_score
ami = adjusted_mutual_info_score(df3['true_label'], df3['predicted_label'])
ari = adjusted_rand_score(df3['true_label'], df3['predicted_label'])
homogeneity = homogeneity_score(df3['true_label'], df3['predicted_label'])
completeness = completeness_score(df3['true_label'], df3['predicted_label'])
vmeasure = v_measure_score(df3['true_label'], df3['predicted_label'])
nmi = normalized_mutual_info_score(df3['true_label'], df3['predicted_label'])


print(f'adjusted_mutual_info_score: {ami}')
print(f'adjusted_rand_score: {ari}')
print(f'normalized_mutual_info_score: {nmi}')
print(f'homogeneity_score: {homogeneity}')
print(f'completeness_score: {completeness}')
print(f'v_measure_score: {vmeasure}')
```

```python
# get the species name associated with the integer value in the dictionary species_dict
species_dict = {v: k for k, v in species_dict.items()}

# display the number of clusters annotated over all sites vs the totalnumber of clusters 
print(f'The total number of clusters that were annotated over all sites is: {len(df3["predicted_label"].unique())} / {df.groupby("site")["cluster_number"].nunique().sum()}')

# display the number of clusters per site
for site in df3['site'].unique():
    print(f'The total number of clusters that were annotated in {site} is: {len(df3.loc[df3["site"] == site, "predicted_label"].unique())} / {df.loc[df["site"] == site, "cluster_number"].nunique()}')

```

```python
# For each cluster, print the species with the number of points per species
for pseudo_label_id in df3['predicted_label'].unique() :
    if pseudo_label_id != -1 :
        print(f'Cluster {pseudo_label_id} :')
        for true_label_id in df3['true_label'].unique() :
            if np.sum((df3['predicted_label'] == pseudo_label_id) & (df3['true_label'] == true_label_id)) > 0:
                print(f'    {species_dict[true_label_id]} : {np.sum((df3["predicted_label"] == pseudo_label_id) & (df3["true_label"] == true_label_id))}')
```

```python
# list of labels corresponding to the most represented species in each cluster
label_to_cluster = []

# For each cluster, select the species with the most points and compute the percentage of points in the cluster
for pseudo_label_id in df3['predicted_label'].unique() :
    if pseudo_label_id != -1 :
        label_to_cluster += [df3['true_label'].unique()[np.argmax([np.sum((df3['predicted_label'] == pseudo_label_id) & (df3['true_label'] == true_label_id)) for true_label_id in np.unique(df3['true_label'])])]]
        print(f'Cluster {pseudo_label_id} : {species_dict[label_to_cluster[-1]]} : {np.round(np.sum((df3["predicted_label"] == pseudo_label_id) & (df3["true_label"] == label_to_cluster[-1])) / np.sum(df3["predicted_label"] == pseudo_label_id) * 100).astype("int")}%')
```

```python
# list of species that are not the most represented in any cluster
missing_species_as_main_cluster = np.setdiff1d(df3['true_label'].unique(), label_to_cluster)

print(f'{len(missing_species_as_main_cluster)} / {len(species_dict)} species are not representing by a cluster.')

print("The list is:")
for label in missing_species_as_main_cluster :
    print(f'    {species_dict[label]}')
```

```python
# for each species in the list missing_species_as_main_cluster, find the clusters where they are present
# and the percentage of points that they represent from the total number of points of the species
for label in missing_species_as_main_cluster :
    print(f'{species_dict[label]} :')
    for pseudo_label_id in np.unique(df3['predicted_label']) :
        if pseudo_label_id != -1 :
            if np.sum((df3['predicted_label'] == pseudo_label_id) & (df3['true_label'] == label)) > 0:
                print(f'    Cluster {pseudo_label_id} : {np.round(np.sum((df3["predicted_label"] == pseudo_label_id) & (df3["true_label"] == label)) / np.sum(df3["true_label"] == label) * 100).astype("int")}%')
    # print the sum of the percentage of points that the species represent in the clusters
    print(f'    Total : {np.round(np.sum([np.sum((df3["predicted_label"] == pseudo_label_id) & (df3["true_label"] == label)) / np.sum(df3["true_label"] == label) * 100 for pseudo_label_id in df3["predicted_label"].unique() if pseudo_label_id != -1])).astype("int")}%')
```
